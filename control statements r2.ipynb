```r
#LEAP YEAR
n=as.numeric(readline("enter year"))
if(n%%400==0){
  print("leap year")
}else if(n%%4==0&& n%%100!=0){
  print("leap year")
}else{
  print("not leap year")
}
```

```r
#A VERY BASIC CALCULATOR
a=as.numeric(readline("enter 1st no."))
b=as.numeric(readline("enter 2nd no."))
operator=readline("enter operator(+,-,*,/):")
if(operator=="+"){
  print(a+b)
}else if(operator=="-"){
  print(a-b)
}else if(operator=="*"){
  print(a*b)
}else if(operator=="/"){
  print(a/b)
}else{
  print("invalid")
}
```
