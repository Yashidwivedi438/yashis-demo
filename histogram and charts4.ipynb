```r
#we will learn SCATTERPLOT, IT Shows relationship between 2 numerical variables
hours=c(1,2,3,4,5,6,7)
marks=c(35,43,48,55,65,73,80)
plot(hours,marks)
```

```r
#we will learn point type and size
hours=c(1,2,3,4,5,6,7)
marks=c(35,43,48,55,65,73,80)
plot(hours,marks,
     pch=19,
     cex=1.5)
```

```r
#we will add point() function that adds another point
boys_hours <- c(2, 3, 4, 5, 6)
boys_marks <- c(40, 50, 60, 70, 80)

girls_hours <- c(2, 3, 4, 5, 6)
girls_marks <- c(45, 55, 65, 75, 85)

plot(boys_hours, boys_marks, pch = 19)

points(girls_hours, girls_marks, pch = 17)
```

```r
#all functions together , pch()=point shape,cex()=controls point size ,xlim()=controls x axis range,grid()=add grid lines to
#graph, abline()=shows trend using straight line , lm()=calculates relationship ,text()=add text near point
hours <- c(1, 2, 3, 4, 5, 6, 7)
marks <- c(35, 42, 48, 55, 65, 72, 80)

plot(hours, marks,
     main = "Study Hours vs Marks",
     xlab = "Study Hours",
     ylab = "Marks",
     pch = 19,
     cex = 1.3,
     xlim = c(0, 8),
     ylim = c(0, 100))

grid()

abline(lm(marks ~ hours))

text(hours, marks, labels = marks, pos = 3)
```

```r
#now we will learn PIE CHARTS
students=c(40,30,20,10)
courses=c("bca","bba","ba","b.com")
pie(students)
```

```r
#now we will learn PIE CHARTS with labels
students=c(40,30,20,10)
courses=c("bca","bba","ba","b.com")
pie(students,
    labels=courses,
    main="student by courses")
```

```r
#now we will learn PIE CHARTS adding colours and labels
students=c(40,30,20,10)
courses=c("bca","bba","ba","b.com")
pie(students,
    labels=courses,
    main="student by courses",
    col=c("red","blue","green","yellow"))
```

```r
#we will learn PIE CHARTS with percentage
students=c(49,32,26,17)
courses=c("bca","bba","ba","b.com")
percentage=round(students/sum(students)*100)
percentage
pie(students,
    labels=courses,
    main="student by courses",
    col=c("red","blue","green","yellow"))
```

```r
#now we will learn PIE CHARTS showing percentage
students=c(40,30,20,10)
courses=c("bca","bba","ba","b.com")
labels=paste(courses,percentage,"%")
pie(students,
    labels=courses,
    main="student by courses",
    col=c("red","blue","green","yellow"))
```

```r
#now we will learn PIE CHARTS changing radius
students=c(40,30,20,10)
courses=c("bca","bba","ba","b.com")
labels=paste(courses,percentage,"%")
pie(students,
    labels=courses,
    radius=1,
    main="student by courses",
    col=c("skyblue","black","white","yellow"))
```
