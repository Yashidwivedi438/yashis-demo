```r
#LOOP using loop we print a certain range of numbers
for(i in 1:5){
    print(i)
}
```

```r
#SUM OF NUMBERS
sum=0
for(i in 1:10) {
  sum=sum+i
}
print(sum)
```

```r
#MULTIPLICATION TABLE
for(i in 1:10){
  print(5*i)
}
```

```r
#TABLE OF A NUMBER GIVEN BY USER
n=as.numeric(readline("enter number"))
for(i in 1:10){
  print(n*i)
}
```

```r
#FACTORIAL
n <- 5
fact <- 1

for(i in 1:n) {
  fact <- fact * i
}

print(fact)
```

```r
#WHILE LOOP
i <- 1

while(i <= 5) {
  print(i)
  i <- i + 1
}
```

```r
#REPEAT LOOP
i <- 1

repeat {
  print(i)
  i <- i + 1

  if(i > 5) {
    break
  }
}
```

```r
#BREAK LOOP
for(i in 1:10) {
  if(i == 6) {
    break
  }
  print(i)
}
```

```r
#next loop
for(i in 1:5) {
  if(i == 3) {
    next
  }
  print(i)
}
```

```r
#nested loop
for(i in 1:3) {
  for(j in 1:3) {
    print(paste(i, j))
  }
}
```
